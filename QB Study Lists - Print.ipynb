```python
import pandas as pd
import pickle

# Load study lists
with open('data/study_lists.pkl', 'rb') as f:
    study_categories = pickle.load(f)

print("✅ Study lists loaded!")
```

## Print Format Options

Choose your preferred format below.

```python
def print_compact_list(category_name, words, max_words=None, min_difficulty=0):
    """
    Compact format - good for printing on paper.
    Shows: difficulty, alphagram, word(s)
    """
    words_filtered = [w for w in words if w['difficulty'] >= min_difficulty]
    words_to_print = words_filtered[:max_words] if max_words else words_filtered
    
    print(f"{'='*50}")
    print(f"{category_name.replace('_', ' ').upper()}")
    if min_difficulty > 0:
        print(f"(Difficulty >= {min_difficulty})")
    print(f"{'='*50}")
    
    for i, entry in enumerate(words_to_print, 1):
        word = entry['word']
        alphagram = entry['alphagram']
        all_words = entry['alphagram_words']
        difficulty = entry['difficulty']
        
        if len(all_words) == 1:
            print(f"{i:3d}. [{difficulty:5.2f}] {alphagram:8s} → {word}")
        else:
            other_words = [w for w in all_words if w != word]
            print(f"{i:3d}. [{difficulty:5.2f}] {alphagram:8s} → {word:12s} (also: {', '.join(other_words)})")
    
    if max_words and len(words_filtered) > max_words:
        print(f"... and {len(words_filtered) - max_words} more")
    
    print(f"Total: {len(words_filtered)} words")

# Example usage
print_compact_list('singleton_4letter', study_categories['singleton_4letter'], max_words=20, min_difficulty=0.25)
```

```python
def print_alphagram_only(category_name, words, max_words=None, min_difficulty=0):
    """
    Alphagram-only format - for pure practice (answers hidden).
    Shows only alphagrams in columns.
    """
    words_filtered = [w for w in words if w['difficulty'] >= min_difficulty]
    words_to_print = words_filtered[:max_words] if max_words else words_filtered
    
    print(f"{'='*50}")
    print(f"{category_name.replace('_', ' ').upper()} - ALPHAGRAMS ONLY")
    if min_difficulty > 0:
        print(f"(Difficulty >= {min_difficulty})")
    print(f"{'='*50}")
    
    # Print in 4 columns
    alphagrams = [entry['alphagram'] for entry in words_to_print]
    col_width = 12
    cols = 4
    
    for i in range(0, len(alphagrams), cols):
        row = alphagrams[i:i+cols]
        row_str = ''.join([f"{a:<{col_width}}" for a in row])
        print(f"{i+1:3d}. {row_str}")
    
    print(f"Total: {len(alphagrams)} alphagrams")

# Example usage
print_alphagram_only('multi_5plus', study_categories['multi_5plus'], max_words=40, min_difficulty=1.0)
```

```python
def print_detailed_list(category_name, words, max_words=None, min_difficulty=0):
    """
    Detailed format with all metadata.
    Shows: difficulty, missed count, last_words avg, frequency, alphagram, word(s)
    """
    words_filtered = [w for w in words if w['difficulty'] >= min_difficulty]
    words_to_print = words_filtered[:max_words] if max_words else words_filtered
    
    print(f"{'='*70}")
    print(f"{category_name.replace('_', ' ').upper()} - DETAILED")
    if min_difficulty > 0:
        print(f"(Difficulty >= {min_difficulty})")
    print(f"{'='*70}")
    print(f"{'#':>3} {'Diff':>6} {'Miss':>5} {'Last':>5} {'Freq':>8} {'Alpha':>8} {'Word(s)'}")
    print("-" * 70)
    
    for i, entry in enumerate(words_to_print, 1):
        word = entry['word']
        alphagram = entry['alphagram']
        all_words = entry['alphagram_words']
        difficulty = entry['difficulty']
        missed = entry['missed_count']
        last_avg = entry['last_words_avg']
        freq = entry['frequency'] if entry['frequency'] else 0
        
        if len(all_words) == 1:
            words_str = word
        else:
            words_str = f"{word} + {len(all_words)-1} more"
        
        print(f"{i:3d} {difficulty:6.2f} {missed:5.2f} {last_avg:5.2f} {freq:8.0f} {alphagram:>8s} {words_str}")
    
    if max_words and len(words_filtered) > max_words:
        print(f"... and {len(words_filtered) - max_words} more")
    
    print(f"Total: {len(words_filtered)} words")

# Example usage
print_detailed_list('singleton_5plus', study_categories['singleton_5plus'], max_words=30, min_difficulty=1.5)
```

## Generate All Lists for Printing

```python
# Customize these settings
MAX_WORDS_PER_CATEGORY = 40  # How many words to print per category (top N by difficulty)
MIN_DIFFICULTY = 0.0  # Only show words with difficulty >= this threshold

categories = [
    ('singleton_4letter', 'Singleton 4-Letter Words'),
    ('multi_4letter', 'Multi-Word 4-Letter Alphagrams'),
    ('singleton_5plus', 'Singleton 5+ Letter Words'),
    ('multi_5plus', 'Multi-Word 5+ Letter Alphagrams'),
]

for cat_key, cat_name in categories:
    print_compact_list(cat_name, study_categories[cat_key], 
                      max_words=MAX_WORDS_PER_CATEGORY, 
                      min_difficulty=MIN_DIFFICULTY)
    print()
```

## Export to Text File

Save to a text file for easy printing.

```python
import sys
from datetime import datetime

# Redirect print to file
output_file = f'study_lists_{datetime.now().strftime("%Y%m%d")}.txt'
with open(output_file, 'w') as f:
    original_stdout = sys.stdout
    sys.stdout = f
    
    print("QUEEN BEE STUDY LISTS")
    print(f"Generated: {datetime.now().strftime('%Y-%m-%d %H:%M')}")
    print(f"Minimum Difficulty: {MIN_DIFFICULTY}")
    print()
    
    for cat_key, cat_name in categories:
        print_compact_list(cat_name, study_categories[cat_key], 
                          max_words=MAX_WORDS_PER_CATEGORY, 
                          min_difficulty=MIN_DIFFICULTY)
        print("\f")  # Form feed - page break for printing
    
    sys.stdout = original_stdout

print(f"✅ Study lists saved to {output_file}")
print(f"   Ready to print!")
```
